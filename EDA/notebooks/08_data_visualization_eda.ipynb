Mục tiêu: Kể lại câu chuyện dữ liệu (Data Storytelling) thông qua các biểu đồ đắt giá nhất.


```python
import runpy
from IPython.display import Image, display

from pathlib import Path

def find_project_root(start=None):
    start = Path(start or Path.cwd()).resolve()
    for path in [start, *start.parents]:
        if (path / 'EDA').exists() and (path / 'ETL').exists():
            return path
    raise FileNotFoundError('Khong tim thay thu muc goc du an chua EDA/ va ETL/.')

PROJECT_ROOT = find_project_root()
EDA_DIR = PROJECT_ROOT / 'EDA'
ETL_DIR = PROJECT_ROOT / 'ETL'

runpy.run_path(str(EDA_DIR / 'src' / '08_data_visualization.py'), run_name='__main__')
IMAGE_DIR = EDA_DIR / 'outputs' / 'figures' / '08_visualization'
```
