# Module 09: Nâng chất lượng dữ liệu (Data Quality)

Mục tiêu: Chứng minh độ sạch của dữ liệu (Data Audit) và chốt quy tắc xử lý ngoại lệ.


```python
import pandas as pd
from IPython.display import display

from pathlib import Path

def find_project_root(start=None):
    start = Path(start or Path.cwd()).resolve()
    for path in [start, *start.parents]:
        if (path / 'EDA').exists() and (path / 'ETL').exists():
            return path
    raise FileNotFoundError('Khong tim thay thu muc goc du an chua EDA/ va ETL/.')

PROJECT_ROOT = find_project_root()
EDA_DIR = PROJECT_ROOT / 'EDA'
ETL_DIR = PROJECT_ROOT / 'ETL'

dq = pd.read_csv(EDA_DIR / 'outputs' / 'tables' / '09_data_quality' / 'data_quality_issues.csv')
display(dq)
```

### Phân tích độ tin cậy của dữ liệu (Data Reliability):
Sau quá trình Audit, kết quả cho thấy:
- **Trùng lặp (Duplicates)**: 0 lỗi. Dữ liệu đã được hệ thống Open University làm sạch rất tốt.
- **Khóa ngoại (Foreign Keys)**: 0 lỗi mồ côi (Orphan records). Mọi bảng (từ VLE, Assessment đến Info) đều trỏ đúng về khóa chính của nhau.

-> **Kết luận**: Sự toàn vẹn dữ liệu được đảm bảo 100%, bộ data hoàn toàn đủ tiêu chuẩn khắt khe để đưa vào thiết kế mô hình Data Warehouse (Star Schema).

### Quy tắc xử lý ngoại lệ được áp dụng:
1. **Về Outliers (Giá trị ngoại lai)**: Các lượt click cao đột biến (hàng chục nghìn lượt) là hành vi học tập có thật của nhóm sinh viên xuất sắc (Tín hiệu thực). Tuyệt đối **KHÔNG** được dùng Z-score hay IQR để cắt bỏ những dòng này.
2. **Về Missing Values**: Thiếu điểm `score` mang ý nghĩa nghiệp vụ là sinh viên chưa có điểm (sẽ được gắn cờ `is_missing_score`). Tương tự, thiếu `date_unregistration` được hiểu đơn giản là sinh viên đó không hề rút lui khóa học.
