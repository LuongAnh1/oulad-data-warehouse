# Module 07: Chuẩn hóa kiểu dữ liệu

Mục tiêu: Làm rõ sự dịch chuyển và lý do ép kiểu dữ liệu từ Raw -> ETL -> EDA.


```python
import runpy
import pandas as pd
from IPython.display import display

from pathlib import Path

def find_project_root(start=None):
    start = Path(start or Path.cwd()).resolve()
    for path in [start, *start.parents]:
        if (path / 'EDA').exists() and (path / 'ETL').exists():
            return path
    raise FileNotFoundError('Khong tim thay thu muc goc du an chua EDA/ va ETL/.')

PROJECT_ROOT = find_project_root()
EDA_DIR = PROJECT_ROOT / 'EDA'
ETL_DIR = PROJECT_ROOT / 'ETL'

runpy.run_path(str(EDA_DIR / 'src' / '07_standardize_dtypes.py'), run_name='__main__')
df = pd.read_csv(EDA_DIR / 'outputs' / 'tables' / '07_standardize_dtypes' / 'dtype_mapping_raw_etl_eda.csv')
display(df)
```

### Phân tích kết quả chuẩn hóa:
1. **Nhóm ID**: Các cột `id_student`, `id_site`, `id_assessment` bắt buộc phải được quy hoạch về dạng `string`. Nếu để số nguyên `int`, mô hình Machine Learning hoặc báo cáo tính tổng/trung bình sẽ hiểu lầm đây là giá trị số học (gây sai lệch phương sai).
2. **Nhóm Phân loại (Categorical)**: Các biến như `code_module`, `final_result` được ép về dạng `category`. Việc này giúp tối ưu RAM đến 80% đối với bảng có hàng triệu dòng như `studentVle`, đồng thời tăng tốc độ xử lý khi GroupBy.
3. **Nhóm Số học (Numeric)**: Các giá trị ngày tháng của OULAD (như `date`, `date_submitted`) thực chất là ngày tương đối (relative days) tính từ thời điểm bắt đầu môn học, do đó hoàn toàn đúng đắn khi giữ nguyên kiểu số nguyên.
