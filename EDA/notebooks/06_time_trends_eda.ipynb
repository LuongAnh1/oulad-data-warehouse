# EDA 06 - Phân tích mẫu và xu hướng theo thời gian

Notebook này dùng để chạy lại script phân tích xu hướng theo thời gian và xem nhanh các bảng, biểu đồ trọng tâm. Nội dung chính được sinh tự động ở `EDA/docs/06-time-trends/`.

```python
from pathlib import Path
import subprocess
import sys

import pandas as pd
from IPython.display import Image, display

cwd = Path.cwd().resolve()
ROOT = cwd.parents[1] if cwd.name == "notebooks" else cwd
SCRIPT = ROOT / "EDA" / "src" / "06_analyze_time_trends.py"
TABLE_DIR = ROOT / "EDA" / "outputs" / "tables" / "06_time_trends"
FIGURE_DIR = ROOT / "EDA" / "outputs" / "figures" / "06_time_trends"
```

## Chạy lại phân tích

```python
subprocess.run([sys.executable, str(SCRIPT)], check=True)
```

## Bảng trọng tâm

```python
weekly_overall = pd.read_csv(TABLE_DIR / "weekly_overall_trends.csv")
weekly_overall.head()
```

```python
trend_change_points = pd.read_csv(TABLE_DIR / "trend_change_points_by_final_result.csv")
trend_change_points
```

```python
checkpoint_metrics = pd.read_csv(TABLE_DIR / "checkpoint_metrics_by_final_result.csv")
checkpoint_metrics
```

## Biểu đồ trọng tâm

```python
for name in [
    "weekly_total_click_line.png",
    "weekly_active_rate_by_final_result_line.png",
    "weekly_click_per_active_by_final_result_line.png",
    "assessment_window_click_by_final_result_line.png",
    "checkpoint_active_rate_by_final_result_line.png",
]:
    display(Image(filename=str(FIGURE_DIR / name)))
```
