# EDA - Phân bố dữ liệu

Notebook này là điểm vào cho phần phân tích phân bố dữ liệu. Logic chính nằm trong `EDA/src/02_analyze_distributions.py` để có thể chạy lại ổn định bằng script.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parents[1]
elif not (ROOT / "EDA").exists():
    ROOT = ROOT.parents[0]

SCRIPT = ROOT / "EDA" / "src" / "02_analyze_distributions.py"
REPORT = ROOT / "EDA" / "docs" / "02-distribution" / "02_distribution_report.md"
FIGURE_DIR = ROOT / "EDA" / "outputs" / "figures" / "02_distribution"
TABLE_DIR = ROOT / "EDA" / "outputs" / "tables" / "02_distribution"

ROOT
```

## Chạy lại phân tích

Cell dưới đây sinh lại toàn bộ bảng, biểu đồ, report và checklist cho mục `Phân bố dữ liệu`.

```python
subprocess.run([sys.executable, str(SCRIPT)], cwd=ROOT, check=True)
```
